# 07 — Frequency Comb Structure

Question: How many frequency channels can one microresonator provide?

\[N_{modes} \approx B/\Delta f\]


```python
from pathlib import Path
import json, numpy as np, pandas as pd, matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

cwd=Path.cwd().resolve()
ROOT=cwd.parent if cwd.name=='notebooks' else cwd
FIGURES_DIR=ROOT/'figures'
CSV_DIR=ROOT/'results'/'csv'
JSON_DIR=ROOT/'results'/'json'
for p in [FIGURES_DIR,CSV_DIR,JSON_DIR]: p.mkdir(parents=True,exist_ok=True)
```

## Mode counting


```python
bandwidth=1000
spacings=np.array([100,50,20,10])
table=pd.DataFrame({'bandwidth':bandwidth,'mode_spacing':spacings,'mode_count':(bandwidth/spacings).astype(int)})
table.to_csv(CSV_DIR/'07_mode_count_table.csv',index=False)
table
```

```python
fig,ax=plt.subplots(figsize=(10,2.5))
ax.hlines(0,0,bandwidth,linewidth=3)
for p in np.linspace(0,bandwidth,21): ax.vlines(p,-0.15,0.15)
ax.set_title('Frequency Comb Across Fixed Bandwidth')
ax.set_xlabel('Frequency')
ax.set_yticks([])
fig.tight_layout(); fig.savefig(FIGURES_DIR/'07_bandwidth_vs_spacing.png',dpi=200)
plt.show()
```

```python
spacing=np.linspace(5,150,200)
modes=bandwidth/spacing
fig,ax=plt.subplots(figsize=(8,5))
ax.plot(spacing,modes,linewidth=2)
ax.set_title('Mode Spacing Tradeoff')
ax.set_xlabel('Mode spacing Δf')
ax.set_ylabel('Available modes')
ax.grid(True,alpha=0.3)
fig.tight_layout(); fig.savefig(FIGURES_DIR/'07_mode_spacing_tradeoff.png',dpi=200)
plt.show()
```

```python
fig,ax=plt.subplots(figsize=(10,3)); ax.set_xlim(0,10); ax.set_ylim(0,5); ax.axis('off')
pump=FancyBboxPatch((0.5,2),1.5,0.8,boxstyle='round,pad=0.02'); ax.add_patch(pump); ax.text(1.25,2.4,'Pump',ha='center')
for i in range(5):
 y=4-i*0.8
 ax.plot([2.0,3.0],[2.4,y],'k-')
 b=FancyBboxPatch((3,y-0.3),2.0,0.6,boxstyle='round,pad=0.02'); ax.add_patch(b); ax.text(4,y,f'Mode {i+1}',ha='center',va='center')
ax.set_title('Addressable Frequency Channels')
fig.tight_layout(); fig.savefig(FIGURES_DIR/'07_addressable_channels.png',dpi=200)
plt.show()
```
